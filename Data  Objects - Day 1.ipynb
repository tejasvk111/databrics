```python
df_accounts = spark.read.option("multiline", True).json("/Volumes/dev/naval/raw/accounts.json")
df_accounts  = df_accounts.withColumn('nominee_name', col('nominee_details.nominee_name'))
df_accounts  = df_accounts.withColumn('relationship', col('nominee_details.relationship'))
df_accounts = df_accounts.drop(df_accounts.nominee_details)
df_accounts = df_accounts.withColumnRenamed('account_id', 'accountid')
df_accounts.write.saveAsTable("accounts_json_formatted")
```
